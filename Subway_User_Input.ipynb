## 2) Prepare Data

```python
from konlpy.tag import Okt

okt = Okt()

def preprocess(sentence):
    tokens = okt.pos(sentence, stem=True)

    allowed_pos = ["Noun", "Verb", "Adjective"]

    filtered = [
        word for word, pos in tokens
        if pos in allowed_pos and len(word) > 1
    ]

    return " ".join(filtered)

#예시
print(preprocess("단백질 많은 메뉴로 추천해주세요"))

```
